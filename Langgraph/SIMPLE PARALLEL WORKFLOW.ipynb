```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class Batsman(TypedDict):
    runs:int
    balls:int
    fours:int
    sixes:int
    sr:float
    bpb:float
    boundary_percent:float
    summary:str
    
```

```python
def calculate_sr(state:Batsman)->Batsman:
    sr=(state['runs']/state['balls'])*100
    state['sr']=sr
    return {'sr':sr}
```

```python
def calculate_bpb(state:Batsman)->Batsman:
    bpb=state['balls']/(state['fours']+state['sixes'])
    state['bpb']=bpb
    return {'bpb':bpb}
```

```python
def calculate_boundary_percent(state:Batsman)->Batsman:
    boundary_percent=(((state['fours']*4)+(state['sixes']*6))/state['runs'])*100
    state['boundary_percent']=boundary_percent
    return {'boundary_percent':boundary_percent}
```

```python
def summary(state:Batsman)->Batsman:
    summary=f"""Strike rate-{state['sr']}\n
    Balls per boundary={state['bpb']}\n
    Boundary percent={state['boundary_percent']}\n"""
    return {'summary': summary}
```

```python
graph=StateGraph(Batsman)
graph.add_node("calculate_sr",calculate_sr)
graph.add_node('calculate_bpb',calculate_bpb)
graph.add_node('calculate_boundary_percent',calculate_boundary_percent)
graph.add_node('summary',summary)
graph.add_edge(START,'calculate_sr')
graph.add_edge(START,'calculate_bpb')
graph.add_edge(START,'calculate_boundary_percent')
graph.add_edge('calculate_sr','summary')
graph.add_edge('calculate_bpb','summary')
graph.add_edge('calculate_boundary_percent','summary')
graph.add_edge('summary',END)
workflow=graph.compile()
graph.compile()
```

```python
initial_state={
    'runs':100,
    'balls':50,
    'fours':6,
    'sixes':4
}
workflow.invoke(initial_state)
```

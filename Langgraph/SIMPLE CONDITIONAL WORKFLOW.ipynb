```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict,Literal
```

```python
class QuadState(TypedDict):
    a:int
    b:int
    c:int
    equation:str
    discriminant:float
    result:str
```

```python
def show_equation(state:QuadState)->QuadState:
    equation=f'{state["a"]}x^2{state["b"]}x{state["c"]}'
    return {"equation":equation}
def calc_d(state:QuadState)->QuadState:
    discriminant= (((state["b"])**2)-4*(state["a"]*state["c"]))
    return {'discriminant':discriminant}
```

```python
def real_root(state:QuadState)->QuadState:
    root1=(-state['b'] + state['discriminant']**0.5) / (2 * state['a'])
    root2=(-state['b'] - state['discriminant']**0.5) / (2 * state['a'])
    result=f'root1={root1} and root2={root2}'
    return {"result":result}
def repeated_root(state:QuadState)->QuadState:
    root=(-state['b']/(2*state['a']))
    result=f'the repeated root is {root}'
    return {'result':result}
def no_real_root(state:QuadState)->QuadState:
    return {'result':f'No real roots exist'}
```

```python
def check_condition(state:QuadState)->Literal['real_root','repeated_root','no_real_root']:
    if state['discriminant']>0:
        return 'real_root'
    elif state['discriminant']<0:
        return 'no_real_root'
    else:
        return "repeated_root"
```

```python
graph=StateGraph(QuadState)
graph.add_node('show_equation',show_equation)
graph.add_node('calculate_discriminant',calc_d)
graph.add_edge(START,'show_equation')
graph.add_edge('show_equation','calculate_discriminant')
graph.add_node("real_root",real_root)
graph.add_node("repeated_root",repeated_root)
graph.add_node("no_real_root",no_real_root)
graph.add_conditional_edges('calculate_discriminant',check_condition)
workflow=graph.compile()
graph.compile()
```

```python
initial_state={
    'a':2,
    'b':4,
    'c':4
}
workflow.invoke(initial_state)
```
